# Generate parameters for simulations

```python
import sys
sys.path.append('../../cogwheel/')
sys.path.append('..')

import warnings
warnings.filterwarnings("ignore", "Wswiglal-redir-stdio")
import lal
lal.swig_redirect_standard_output_error(False)

from pathlib import Path

from cogwheel_machine import training_priors
from cogwheel_machine import config
```

```python
prior = training_priors.NoSpinTrainingPrior(**config.PRIOR_KWARGS)
```

```python
n_samples = 10**2
simulation_parameters = prior.generate_random_samples(n_samples)
```

```python
sim_dir = Path('../data/set_0')
sim_dir.mkdir(parents=True, exist_ok=True)
simulation_parameters.to_feather(sim_dir/'simulation_parameters.feather')
```
